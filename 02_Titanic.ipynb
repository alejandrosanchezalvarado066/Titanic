## Introducción

En 1912, el personal encargado del embarque del Titanic utilizaba listas en papel para registrar a los pasajeros. En esta práctica, estudiaremos ese registro como un ejemplo de procesamiento manual de transacciones. Las listas cumplían una función operativa: dejar constancia de quiénes viajaban. El dataset reúne datos de los pasajeros, como su clase y puerto de embarque, e incorpora un dato conocido después del naufragio: si sobrevivieron.

### Diccionario de datos

El archivo `titanic.csv` contiene **891 registros y 12 columnas**. Cada fila representa a un pasajero.

| Columna | Tipo de dato | Descripción |
|---|---|---|
| `PassengerId` | Identificador entero | Código único asignado a cada pasajero en el dataset. |
| `Survived` | Categoría binaria | Resultado registrado después del naufragio. `0` = falleció, `1` = sobrevivió. |
| `Pclass` | Categoría ordinal | Clase del boleto. `1` = primera, `2` = segunda, `3` = tercera. |
| `Name` | Texto | Nombre del pasajero tal como aparece en el archivo. |
| `Sex` | Categoría | Sexo registrado. Los valores del archivo son `female` y `male`. |
| `Age` | Número | Edad en años. Puede incluir valores fraccionarios, por ejemplo, para bebés menores de un año. |
| `SibSp` | Conteo entero | Número de hermanos, hermanas y cónyuges del pasajero a bordo. |
| `Parch` | Conteo entero | Número de padres, madres, hijos e hijas del pasajero a bordo. |
| `Ticket` | Código de texto | Código del boleto. Puede incluir letras, números y signos, y repetirse en los registros de varios pasajeros. |
| `Fare` | Número decimal | Tarifa registrada en la fila del pasajero. |
| `Cabin` | Código de texto | Identificación de la cabina, cuando figura en el registro. |
| `Embarked` | Categoría | Puerto de embarque. `C` = Cherburgo, `Q` = Queenstown, `S` = Southampton. |

```python
#acceder al ccv
import pandas as pd

df = pd.read_csv("titanic.csv")
#print (df)

#1. ¿Cuántos pasajeros hay en el dataset 
# y cuántos PassengerId distintos aparecen? ¿Coinciden los dos resultados?

n_pasajeros = len(df)
n_Iddistinto = len(df["PassengerId"].unique())

print("numero de pasajero: ", n_pasajeros)
print("numero de iDdistinto: ", n_Iddistinto)
print("=============================")
```

```python

#2. ¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?
print ("Primea clase: ", len(df[df ["Pclass"] == 1]) )
print ("Segunda clase: ", len(df[df ["Pclass"] == 2]) )
print ("Tercera clase: ", len(df[df ["Pclass"] == 3]) )

print("=============================")

```

```python

#3. ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?
print ("cherburgo: ", len(df[df ["Embarked"] == "C"]) )
print ("Queenston: ", len(df[df ["Embarked"] == "Q"]) )
print ("Southampton: ", len(df[df ["Embarked"] == "S"]) )
print("Sin embarque:", len(df[df["Embarked"].isnull()]))
print("=============================")
```

```python

#4. ¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?
print ("Primera clase: ")
print("Tarifa promedio: ", df[df["Pclass"] == 1] ["Fare"].mean())
print("Tarifa max: ", df[df["Pclass"] == 1] ["Fare"].max())
print("Tarifa min: ", df[df["Pclass"] == 1] ["Fare"].min())

print ("Segunda clase: ")
print("Tarifa promedio: ", df[df["Pclass"] == 2] ["Fare"].mean())
print("Tarifa max: ", df[df["Pclass"] == 2] ["Fare"].max())
print("Tarifa min: ", df[df["Pclass"] == 2] ["Fare"].min())

print ("Tercera clase: ")
print("Tarifa promedio: ", df[df["Pclass"] == 3] ["Fare"].mean())
print("Tarifa max: ", df[df["Pclass"] == 3] ["Fare"].max())
print("Tarifa min: ", df[df["Pclass"] == 3] ["Fare"].min())
print("=============================")
```

```python

#5. ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?
print ("Primera clase: ")
pClase = df[df["Pclass"] == 1]
print ("cherburgo", len (pClase [pClase["Embarked"] == "C"]))
print ("Queenston", len (pClase [pClase["Embarked"] == "Q"]))
print ("Southampton", len (pClase [pClase["Embarked"] == "S"]))
print ("Sin embarque", len (pClase [pClase["Embarked"].isnull()]))

print ("Segunda clase: ")
sClase = df[df["Pclass"] == 2]
print ("cherburgo", len (sClase [sClase["Embarked"] == "C"]))
print ("Queenston", len (sClase [sClase["Embarked"] == "Q"]))
print ("Southampton", len (sClase [sClase["Embarked"] == "S"]))
print ("Sin embarque", len (sClase [sClase["Embarked"].isnull()]))

print ("Tercera clase: ")
tClase = df[df["Pclass"] == 3]
print ("cherburgo", len (tClase [tClase["Embarked"] == "C"]))
print ("Queenston", len (tClase [tClase["Embarked"] == "Q"]))
print ("Southampton", len (tClase [tClase["Embarked"] == "S"]))
print ("Sin embarque", len (tClase [tClase["Embarked"].isnull()]))
print("=============================")

```

```python

#6. ¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?
print("Primera clase: ")
print("Sobrevivieron: ", len(pClase[pClase["Survived"] == 1]))
print("Murieron: ", len(pClase[pClase["Survived"] == 0]))

print("Segunda clase: ")
print("Sobrevivieron: ", len(sClase[sClase["Survived"] == 1]))
print("Murieron: ", len(sClase[sClase["Survived"] == 0]))

print("Tercera clase: ")
print("Sobrevivieron: ", len(tClase[tClase["Survived"] == 1]))
print("Murieron: ", len(tClase[tClase["Survived"] == 0]))
print("=============================")

```

```python
#7. ¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?
print ("Cero familiares: ", len (df[(df["SibSp"] == 0) & (df["Parch"] == 0)]))
print ("Un familiar", len (df[ df["SibSp"]+df["Parch"] == 1  ]))
print ("dos familiar", len (df[ df["SibSp"]+df["Parch"] == 2  ]))
print ("mas de dos familiar", len (df[ df["SibSp"]+df["Parch"] > 2  ]))

```

```python
#8. ¿Cuántos pasajeros tenían menos de 18 años, 
# cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.
print ("Primera CLase: ")
print("menor 18 años: ", len(pClase[pClase["Age"] < 18]))
print("mayor o igual 18 años: ", len(pClase[pClase["Age"] >= 18]))
print("Sin registrar edad: ", len(pClase[ pClase["Age"].isnull() ]))

print ("Segunda CLase: ")
print("menor 18 años: ", len(sClase[sClase["Age"] < 18]))
print("mayor o igual 18 años: ", len(sClase[sClase["Age"] >= 18]))
print("Sin registrar edad: ", len(sClase[ sClase["Age"].isnull() ]))

print ("Tercera CLase: ")
print("menor 18 años: ", len(tClase[tClase["Age"] < 18]))
print("mayor o igual 18 años: ", len(tClase[tClase["Age"] >= 18]))
print("Sin registrar edad: ", len(tClase[ tClase["Age"].isnull() ]))

```

```python

#9. ¿Cuántos códigos de boleto distintos hay? 
# ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?
print ("Codigos distintos: ", len(df["Ticket"].unique()))
print ("Codigos repetidos: ", len(df["Ticket"].duplicated()))
print ("El mas repetido es: ", df["Ticket"].value_counts().index[0], " se repite: ", len (df[ df["Ticket"] == df["Ticket"].value_counts().index[0]]))


```

```python
#10. ¿En qué clase hay más registros sin cabina? 
# Para cada clase, muestren cuántos registros tienen cabina, 
# cuántos carecen de ella y cuántos hay en total.

n = df[df["Cabin"].isnull()].groupby("Pclass").size().max()
c = df[df["Cabin"].isnull()].groupby("Pclass").size().idxmax()

print ("La clase con mas registro sin cabina es la clase: ", c, " con: ", n, " pasajeros" )

for i in [1,2,3]:
    clase = df[df["Pclass"] == i]
    print ("En la clase: ", i)
    print ("Sin cabina: ", clase[clase["Cabin"].isnull()].groupby("Pclass").size().max())
    print ("Con cabina: ", clase[clase["Cabin"].notnull()].groupby("Pclass").size().max())
    print (" Total: ", len(clase))
    print("==========================")
    
    
     

```
